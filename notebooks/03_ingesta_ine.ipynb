```python
from pyspark.sql import functions as F

ruta_tasas = "/Volumes/workspace/default/tfm_visnu_raw/raw/Tasas de actividad, paro y empleo por provincia y sexo.csv"
ruta_renta = "/Volumes/workspace/default/tfm_visnu_raw/raw/Renta neta media por hogar y por persona.csv"
ruta_pob = "/Volumes/workspace/default/tfm_visnu_raw/raw/Población según comunidad autónoma y provincia y sexo.csv"
```

```python
display(dbutils.fs.ls("/Volumes/workspace/default/tfm_visnu_raw/raw/"))
```

```python
df_tasas = (
    spark.read
    .option("header", True)
    .option("sep", ";")
    .option("encoding", "utf-8")
    .csv(ruta_tasas)
)

display(df_tasas)
```

```python
print(df_tasas.columns)
```

```python
##limpieza de tasas

df_tasas_clean = (
    df_tasas
    .withColumn("provincia_raw", F.col("Provincias"))
    .withColumn("sexo", F.col("Sexo"))
    .withColumn("tipo_tasa", F.col("Tasas"))
    .withColumn("periodo", F.col("Periodo"))
    .withColumn("valor_str", F.trim(F.col("Total")))
    .withColumn("valor_str", F.regexp_replace("valor_str", r"\.", ""))
    .withColumn("valor_str", F.regexp_replace("valor_str", ",", "."))
    .withColumn(
        "valor",
        F.when(
            (F.col("valor_str").isNull()) |
            (F.col("valor_str") == "") |
            (F.col("valor_str") == "."),
            None
        ).otherwise(F.col("valor_str").cast("double"))
    )
    .filter(F.col("provincia_raw").isNotNull())
    .filter(~F.col("provincia_raw").contains("Total Nacional"))
    .withColumn("anio", F.regexp_extract("periodo", r"(\d{4})", 1).cast("int"))
    .withColumn("provincia", F.regexp_replace("provincia_raw", r"^\d+\s+", ""))
)
```

```python
display(
    df_tasas_clean
    .filter(F.col("valor").isNull())
    .select("Sexo", "Provincias", "Tasas", "Periodo", "Total", "valor_str")
)
```

```python
##normalizo provincias

df_tasas_clean = (
    df_tasas_clean
    .withColumn("provincia", F.upper(F.trim(F.col("provincia"))))
    .withColumn("provincia", F.regexp_replace("provincia", "Á", "A"))
    .withColumn("provincia", F.regexp_replace("provincia", "É", "E"))
    .withColumn("provincia", F.regexp_replace("provincia", "Í", "I"))
    .withColumn("provincia", F.regexp_replace("provincia", "Ó", "O"))
    .withColumn("provincia", F.regexp_replace("provincia", "Ú", "U"))
    .withColumn("provincia", F.regexp_replace("provincia", "Ü", "U"))
    .withColumn("provincia", F.regexp_replace("provincia", "Ñ", "N"))
)

display(df_tasas_clean)
```

```python
df_tasas_anual = (
    df_tasas_clean
    .groupBy("anio", "provincia", "sexo", "tipo_tasa")
    .agg(F.avg("valor").alias("valor_medio_anual"))
)

display(df_tasas_anual)
```

```python
df_tasas_pivot = (
    df_tasas_anual
    .withColumn(
        "variable",
        F.concat_ws("_", F.col("tipo_tasa"), F.col("sexo"))
    )
    .groupBy("anio", "provincia")
    .pivot("variable")
    .agg(F.first("valor_medio_anual"))
)

display(df_tasas_pivot)
```

```python
print(df_tasas_pivot.columns)
```

```python
df_tasas_final = (
    df_tasas_pivot
    .withColumnRenamed("Tasa de actividad_Ambos sexos", "tasa_actividad_total")
    .withColumnRenamed("Tasa de actividad_Hombres", "tasa_actividad_hombres")
    .withColumnRenamed("Tasa de actividad_Mujeres", "tasa_actividad_mujeres")
    .withColumnRenamed("Tasa de empleo de la población_Ambos sexos", "tasa_empleo_total")
    .withColumnRenamed("Tasa de empleo de la población_Hombres", "tasa_empleo_hombres")
    .withColumnRenamed("Tasa de empleo de la población_Mujeres", "tasa_empleo_mujeres")
    .withColumnRenamed("Tasa de paro de la población_Ambos sexos", "tasa_paro_total")
    .withColumnRenamed("Tasa de paro de la población_Hombres", "tasa_paro_hombres")
    .withColumnRenamed("Tasa de paro de la población_Mujeres", "tasa_paro_mujeres")
)

display(df_tasas_final)
```

```python
print(df_tasas_final.columns)
```

```python
df_renta = (
    spark.read
    .option("header", True)
    .option("sep", ";")
    .option("encoding", "latin1")
    .csv(ruta_renta)
)

display(df_renta)
```

```python
print(df_renta.columns)
```

```python
df_renta_clean = (
    df_renta
    .withColumn("provincia", F.col("Provincias"))
    .withColumn("indicador", F.col("Indicadores de renta media"))
    .withColumn("anio", F.col("Periodo").cast("int"))
    .withColumn("valor_str", F.trim(F.col("Total")))
    .withColumn("valor_str", F.regexp_replace("valor_str", r"\.", ""))
    .withColumn("valor_str", F.regexp_replace("valor_str", ",", "."))
    .withColumn(
        "valor",
        F.when(
            (F.col("valor_str").isNull()) |
            (F.col("valor_str") == "") |
            (F.col("valor_str") == "."),
            None
        ).otherwise(F.col("valor_str").cast("double"))
    )
    .filter(F.col("provincia").isNotNull())
    .filter(F.trim(F.col("provincia")) != "")
    .filter(~F.col("provincia").contains("Total Nacional"))
    .withColumn("provincia", F.upper(F.trim(F.col("provincia"))))
    .withColumn("provincia", F.regexp_replace("provincia", "Á", "A"))
    .withColumn("provincia", F.regexp_replace("provincia", "É", "E"))
    .withColumn("provincia", F.regexp_replace("provincia", "Í", "I"))
    .withColumn("provincia", F.regexp_replace("provincia", "Ó", "O"))
    .withColumn("provincia", F.regexp_replace("provincia", "Ú", "U"))
    .withColumn("provincia", F.regexp_replace("provincia", "Ü", "U"))
    .withColumn("provincia", F.regexp_replace("provincia", "Ñ", "N"))
)

display(df_renta_clean)
```

```python
display(
    df_renta_clean
    .select("indicador")
    .distinct()
    .orderBy("indicador")
)
```

```python
df_renta_pivot = (
    df_renta_clean
    .groupBy("anio", "provincia")
    .pivot("indicador")
    .agg(F.first("valor"))
)

display(df_renta_pivot)
```

```python
print(df_renta_pivot.columns)
```

```python
df_renta_final = (
    df_renta_pivot
    .withColumnRenamed("Renta neta media por hogar", "renta_media_hogar")
    .withColumnRenamed("Renta neta media por persona", "renta_media_persona")
)

display(df_renta_final)
```

```python
df_pob = (
    spark.read
    .option("header", True)
    .option("sep", ";")
    .option("encoding", "utf-8")
    .csv(ruta_pob)
)

display(df_pob)
```

```python
print(df_pob.columns)
```

```python
df_pob_clean = (
    df_pob
    .withColumn("provincia", F.col("Provincias"))
    .withColumn("sexo", F.col("Sexo"))
    .withColumn("anio", F.col("Periodo").cast("int"))
    .withColumn("valor_str", F.trim(F.col("Total")))
    .withColumn("valor_str", F.regexp_replace("valor_str", r"\.", ""))
    .withColumn("valor_str", F.regexp_replace("valor_str", ",", "."))
    .withColumn(
        "valor",
        F.when(
            (F.col("valor_str").isNull()) |
            (F.col("valor_str") == "") |
            (F.col("valor_str") == "."),
            None
        ).otherwise(F.col("valor_str").cast("double"))
    )
    .filter(F.col("provincia").isNotNull())
    .filter(F.trim(F.col("provincia")) != "")
    .filter(~F.col("provincia").contains("Total Nacional"))
    .withColumn("provincia", F.upper(F.trim(F.col("provincia"))))
    .withColumn("provincia", F.regexp_replace("provincia", "Á", "A"))
    .withColumn("provincia", F.regexp_replace("provincia", "É", "E"))
    .withColumn("provincia", F.regexp_replace("provincia", "Í", "I"))
    .withColumn("provincia", F.regexp_replace("provincia", "Ó", "O"))
    .withColumn("provincia", F.regexp_replace("provincia", "Ú", "U"))
    .withColumn("provincia", F.regexp_replace("provincia", "Ü", "U"))
    .withColumn("provincia", F.regexp_replace("provincia", "Ñ", "N"))
)

display(df_pob_clean)
```

```python
display(
    df_pob_clean
    .select("sexo")
    .distinct()
    .orderBy("sexo")
)
```

```python
df_pob_pivot = (
    df_pob_clean
    .groupBy("anio", "provincia")
    .pivot("sexo")
    .agg(F.first("valor"))
)

display(df_pob_pivot)
```

```python
print(df_pob_pivot.columns)
```

```python
df_pob_final = (
    df_pob_pivot
    .withColumnRenamed("Hombres", "poblacion_hombres")
    .withColumnRenamed("Mujeres", "poblacion_mujeres")
    .withColumnRenamed("Total", "poblacion_total")
)

display(df_pob_final)
```

```python
from pyspark.sql import functions as F

# quitar códigos tipo "01 "
df_pob_final = df_pob_final.withColumn(
    "provincia",
    F.regexp_replace(F.col("provincia"), "^[0-9]{2} ", "")
)
```

```python
def normalizar_provincia(df, col="provincia"):
    return (
        df
        .withColumn(col, F.upper(F.trim(F.col(col))))
        .withColumn(col, F.regexp_replace(col, "Á", "A"))
        .withColumn(col, F.regexp_replace(col, "É", "E"))
        .withColumn(col, F.regexp_replace(col, "Í", "I"))
        .withColumn(col, F.regexp_replace(col, "Ó", "O"))
        .withColumn(col, F.regexp_replace(col, "Ú", "U"))
        .withColumn(col, F.regexp_replace(col, "Ü", "U"))
        .withColumn(col, F.regexp_replace(col, "Ñ", "N"))
    )

df_pob_final = normalizar_provincia(df_pob_final)
df_renta_pivot = normalizar_provincia(df_renta_pivot)
df_tasas_pivot = normalizar_provincia(df_tasas_pivot)
```

```python
df_pob_final = df_pob_final.filter(F.col("anio").between(2021, 2023))
df_renta_pivot = df_renta_pivot.filter(F.col("anio").between(2021, 2023))
df_tasas_pivot = df_tasas_pivot.filter(F.col("anio").between(2021, 2023))
```

```python
print("POBLACION")
display(df_pob_final.selectExpr(
    "count(*) as filas",
    "count(distinct provincia) as provincias",
    "min(anio) as anio_min",
    "max(anio) as anio_max"
))

print("RENTA")
display(df_renta_pivot.selectExpr(
    "count(*) as filas",
    "count(distinct provincia) as provincias",
    "min(anio) as anio_min",
    "max(anio) as anio_max"
))

print("TASAS")
display(df_tasas_pivot.selectExpr(
    "count(*) as filas",
    "count(distinct provincia) as provincias",
    "min(anio) as anio_min",
    "max(anio) as anio_max"
))
```

```python
print("Población sin renta")
display(
    df_pob_final
    .select("anio", "provincia").distinct()
    .join(
        df_renta_pivot.select("anio", "provincia").distinct(),
        on=["anio", "provincia"],
        how="left_anti"
    )
    .orderBy("anio", "provincia")
)

print("Población sin tasas")
display(
    df_pob_final
    .select("anio", "provincia").distinct()
    .join(
        df_tasas_pivot.select("anio", "provincia").distinct(),
        on=["anio", "provincia"],
        how="left_anti"
    )
    .orderBy("anio", "provincia")
)
```

```python
print("Provincias población")
display(df_pob_final.select("provincia").distinct().orderBy("provincia"))

print("Provincias renta")
display(df_renta_pivot.select("provincia").distinct().orderBy("provincia"))

print("Provincias tasas")
display(df_tasas_pivot.select("provincia").distinct().orderBy("provincia"))
```

```python
df_ine_base = (
    df_pob_final
    .join(df_renta_final, on=["anio", "provincia"], how="left")
    .join(df_tasas_final, on=["anio", "provincia"], how="left")
    .orderBy("anio", "provincia")
)

display(df_ine_base)
```

```python
print(df_ine_base.columns)
```

```python
display(
    df_ine_base
    .select("anio")
    .distinct()
    .orderBy("anio")
)
```

```python
display(
    df_ine_base
    .select("provincia")
    .distinct()
    .orderBy("provincia")
)
```

```python
df_ine_base = (
    df_ine_base
    .withColumn("provincia", F.regexp_replace("provincia", r"^\d+\s+", ""))
    .withColumn("provincia", F.upper(F.trim(F.col("provincia"))))
    .withColumn("provincia", F.regexp_replace("provincia", "ARABA/ALAVA", "ALAVA"))
    .withColumn("provincia", F.regexp_replace("provincia", "ALICANTE/ALACANT", "ALICANTE"))
    .withColumn("provincia", F.regexp_replace("provincia", "BALEARS, ILLES", "ILLES BALEARS"))
    .withColumn("provincia", F.regexp_replace("provincia", "CASTELLON/CASTELLO", "CASTELLON"))
    .withColumn("provincia", F.regexp_replace("provincia", "CORUNA, A", "A CORUNA"))
    .withColumn("provincia", F.regexp_replace("provincia", "RIOJA, LA", "LA RIOJA"))
    .withColumn("provincia", F.regexp_replace("provincia", "PALMAS, LAS", "LAS PALMAS"))
    .withColumn("provincia", F.regexp_replace("provincia", "VALENCIA/VALÈNCIA", "VALENCIA"))
)
```

```python
df_ine_base = df_ine_base.filter(F.col("anio").between(2021, 2024))

display(
    df_ine_base
    .select("anio")
    .distinct()
    .orderBy("anio")
)
```
